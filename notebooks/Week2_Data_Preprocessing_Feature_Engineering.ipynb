```python
import pandas as pd

crop = pd.read_csv("hypothetical_crop_yield_data.csv")
weather = pd.read_csv("hypothetical_weather_data.csv")
soil = pd.read_csv("hypothetical_soil_data.csv")

print("Crop dataset loaded:", crop.shape)
print("Weather dataset loaded:", weather.shape)
print("Soil dataset loaded:", soil.shape)
```

```python
print("CROP DATASET")
print("Shape:", crop.shape)
print("Columns:")
print(crop.columns.tolist())

print("\nWEATHER DATASET")
print("Shape:", weather.shape)
print("Columns:")
print(weather.columns.tolist())

print("\nSOIL DATASET")
print("Shape:", soil.shape)
print("Columns:")
print(soil.columns.tolist())
```

```python
print("CROP DATA")
display(crop)

print("WEATHER DATA")
display(weather)

print("SOIL DATA")
display(soil)
```

```python
print("CROP DATA - Missing Values")
print(crop.isnull().sum())

print("\nWEATHER DATA - Missing Values")
print(weather.isnull().sum())

print("\nSOIL DATA - Missing Values")
print(soil.isnull().sum())
```

```python
print("CROP DATA - Duplicate Rows")
print(crop.duplicated().sum())

print("\nWEATHER DATA - Duplicate Rows")
print(weather.duplicated().sum())

print("\nSOIL DATA - Duplicate Rows")
print(soil.duplicated().sum())
```

```python
print("CROP DUPLICATES")
display(crop[crop.duplicated(keep=False)])

print("WEATHER DUPLICATES")
display(weather[weather.duplicated(keep=False)])

print("SOIL DUPLICATES")
display(soil[soil.duplicated(keep=False)])
```

```python
print("Negative Production Values:")
display(crop[crop["Production_Tonnes"] < 0])

print("\nSuspicious Soil pH Values:")
display(soil[(soil["pH"] < 3) | (soil["pH"] > 10)])

print("\nSuspicious Crop Area Values:")
display(crop[crop["Area_Hectares"] > crop["Area_Hectares"].quantile(0.95)])

print("\nExtreme Rainfall Values:")
display(weather[weather["Rainfall_mm"] > weather["Rainfall_mm"].quantile(0.95)])
```

```python
print("CROP STATISTICS")
display(crop.describe())

print("WEATHER STATISTICS")
display(weather.describe())

print("SOIL STATISTICS")
display(soil.describe())
```

```python
# Feature Engineering Examples

crop_fe = crop.copy()
weather_fe = weather.copy()
soil_fe = soil.copy()


crop_fe["Yield_per_Hectare"] = (
    crop_fe["Production_Tonnes"] / crop_fe["Area_Hectares"]
)


crop_fe["Fertilizer_per_Hectare"] = (
    crop_fe["Fertilizer_Kg"] / crop_fe["Area_Hectares"]
)


weather_fe["Temperature_Range"] = (
    weather_fe["Temperature_C"].max() -
    weather_fe["Temperature_C"].min()
)


soil_fe["NPK_Total"] = (
    soil_fe["Nitrogen_mgkg"].fillna(0) +
    soil_fe["Phosphorus_mgkg"].fillna(0) +
    soil_fe["Potassium_mgkg"].fillna(0)
)

print("Crop engineered features:")
display(crop_fe.head())

print("Weather engineered features:")
display(weather_fe.head())

print("Soil engineered features:")
display(soil_fe.head())
```
